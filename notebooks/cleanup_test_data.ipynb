# Clean up generated test data

⚠️ **Warning:** This permanently deletes the test data generated by this repository from the configured bucket.

Only the known generated file paths are deleted. GCS folders are virtual and disappear when their files are removed. 🗑️☁️

```python
import gcsfs

from populate_database import BUCKET, _invalid_files, _valid_paths

confirmation = input("Type 'delete' to permanently delete all generated test data: ")
if confirmation != "delete":
    raise RuntimeError("Cleanup cancelled. Type exactly 'delete' to continue.")

filesystem = gcsfs.GCSFileSystem()
paths = _valid_paths() + _invalid_files()
deleted = 0
for path in paths:
    full_path = f"{BUCKET}/{path}"
    if filesystem.exists(full_path):
        filesystem.rm(full_path)
        deleted += 1
        print(f"[cleanup] Removed {full_path}")

print(f"[cleanup] Removed {deleted} generated files from gs://{BUCKET} ✅")
```
